Rekurencja `T(n) = n + 2*T(n/2)` oraz porownanie z funkcja `O(n) = n + n*log(n)/log(2)`.

```python
import numpy as np
import matplotlib.pyplot as plt


def T(n):
    if n < 2.0:
        return 1.0 + 2.4427 * (n - 1.0) + 0.721348 * (n - 1.0) ** 2
    return n + 2.0 * T(n / 2.0)


n = np.arange(1, 1001)
t_values = np.array([T(float(value)) for value in n])
o_values = n + n * np.log(n) / np.log(2.0)

print("| {0:>5} | {1:>14} | {2:>14} | {3:>10} |".format("n", "T(n)", "O(n)", "T/O"))
print("+-{0:>5}-+-{1:>14}-+-{2:>14}-+-{3:>10}-+".format("-----", "--------------", "--------------", "----------"))
for index in [1, 2, 4, 8, 16, 32, 64, 128, 256, 512, 1000]:
    k = index - 1
    print("| {0:5d} | {1:14.6f} | {2:14.6f} | {3:10.6f} |".format(index, t_values[k], o_values[k], t_values[k] / o_values[k]))

plt.figure(figsize=(10, 5))
plt.plot(n, t_values, label="T(n)")
plt.plot(n, o_values, "--", label="O(n) = n + n log2(n)")
plt.xlabel("n")
plt.ylabel("wartosc")
plt.title("Porownanie T(n) oraz O(n)")
plt.grid(True)
plt.legend()
plt.show()

print("Wniosek: wykresy maja ten sam rzad wzrostu, czyli O(n log n).")
print("Taka klasa odpowiada np. sortowaniu przez scalanie dla danych dzielonych na polowy.")
```
