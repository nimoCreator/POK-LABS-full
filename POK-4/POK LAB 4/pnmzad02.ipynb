importy:

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad
from scipy.special import i0
```

Funkcja impelenentująca całkę Catalana

```python
def Catalan(v):
    value, _ = quad(lambda u: np.exp(v * np.cos(u)), 0.0, np.pi)
    return value / np.pi
```

Kod testowy:

```python

v_points = np.linspace(0.0, 5.0, 11)
integral_values = np.array([Catalan(v) for v in v_points])
bessel_values = i0(v_points)

print("| {0:>7} | {1:>16} | {2:>16} | {3:>13} |".format("v", "I(v)", "i0(v)", "blad"))
print("+-{0:>7}-+-{1:>16}-+-{2:>16}-+-{3:>13}-+".format("-------", "----------------", "----------------", "-------------"))
for v, integral, bessel in zip(v_points, integral_values, bessel_values):
    print("| {0:7.3f} | {1:16.10f} | {2:16.10f} | {3:13.6e} |".format(v, integral, bessel, abs(integral - bessel)))

v_plot = np.linspace(0.0, 5.0, 300)
plt.figure(figsize=(10, 5))
plt.plot(v_plot, i0(v_plot), label="i0(v)")
plt.scatter(v_points, integral_values, color="crimson", label="calka Catalana")
plt.xlabel("v")
plt.ylabel("wartosc")
plt.title("Calka Catalana i funkcja Bessela i0")
plt.grid(True)
plt.legend()
plt.show()
```
