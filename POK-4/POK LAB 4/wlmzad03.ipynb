Pochodna symboliczna funkcji `g(x) = tan(x^3) - 4*tan(x)` oraz ilorazy roznicowe.

```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt


x, h = sp.symbols("x h")
g = sp.tan(x ** 3) - 4 * sp.tan(x)
Dg = sp.diff(g, x)
Ri = (g.subs(x, x + h) - g) / h
limit_Ri = sp.simplify(sp.limit(Ri, h, 0))

print("g(x) =", g)
print("Dg(x) =", Dg)
print("lim Ri =", limit_Ri)
print("Dg == lim Ri:", sp.simplify(Dg - limit_Ri) == 0)

g_num = sp.lambdify(x, g, "numpy")
Dg_num = sp.lambdify(x, Dg, "numpy")
Ri_num = sp.lambdify((x, h), Ri, "numpy")
xs = np.linspace(0.0, 1.0, 500)

plt.figure(figsize=(10, 5))
plt.plot(xs, g_num(xs), label="g(x)")
plt.plot(xs, Dg_num(xs), label="Dg(x)")
plt.xlabel("x")
plt.ylabel("wartosc")
plt.title("Funkcja i pochodna")
plt.ylim(-20, 20)
plt.grid(True)
plt.legend()
plt.show()

plt.figure(figsize=(10, 5))
plt.plot(xs, Dg_num(xs), linewidth=3, label="Dg(x)")
for step in [1.0, 0.5, 0.1, 0.01]:
    plt.plot(xs, Ri_num(xs, step), label="Ri, h = " + str(step))
plt.xlabel("x")
plt.ylabel("wartosc")
plt.title("Pochodna i ilorazy roznicowe")
plt.ylim(-20, 20)
plt.grid(True)
plt.legend()
plt.show()
```
