```python
import numpy as np
import matplotlib.pyplot as plt
from sympy import simplify
```

definicja algorytmu rekurencyjnego T:

```python
def memo(func):
    cache = {}

    def wrapper(n):
        key = round(n, 6)
        if key in cache:
            return cache[key]
        result = func(n)
        cache[key] = result
        return result

    return wrapper


@memo
def T(n):
    if n < 2.0:
        return simplify(1.0 + 2.4427 * (n - 1.0) + 0.721348 * (n - 1.0) ** 2)
    return simplify(n + 2.0 * T(n / 2.0))
```

definicja algorytmu z logarytmem naturalnym:

```python
def O(n):
    return simplify(n + n * np.log(n) / np.log(2.0))
```

główny kod programu:

```python
n = np.arange(1, 1001)
t_values = np.array([T(float(value)) for value in n])
o_values = np.array([O(float(value)) for value in n])

print("| {0:>5} | {1:>14} | {2:>14} | {3:>10} |".format("n", "T(n)", "O(n)", "T/O"))
print("+-{0:>5}-+-{1:>14}-+-{2:>14}-+-{3:>10}-+".format("-----", "--------------", "--------------", "----------"))
for index in [1, 2, 4, 8, 16, 32, 64, 128, 256, 512, 1000]:
    k = index - 1
    print("| {0:5d} | {1:14.6f} | {2:14.6f} | {3:10.6f} |".format(index, t_values[k], o_values[k], t_values[k] / o_values[k]))

plt.figure(figsize=(10, 5))
plt.plot(n, t_values, label="T(n)")
plt.plot(n, o_values, "--", label="O(n) = n + n log2(n)")
plt.xlabel("n")
plt.ylabel("wartosc")
plt.title("Porownanie T(n) oraz O(n)")
plt.grid(True)
plt.legend()
plt.show()

plt.figure(figsize=(10, 5))
plt.plot(n, t_values, label="T(n)")
plt.plot(n, o_values, "--", label="O(n) = n + n log2(n)")
plt.xlabel("n")
plt.ylabel("wartosc")
plt.title("Porownanie T(n) oraz O(n)")
plt.grid(True)
plt.legend()
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```

Wygląda na to że obie funkcje dążą doi nieskończoności w kategorii O(n), czli funkcji liniowej.
