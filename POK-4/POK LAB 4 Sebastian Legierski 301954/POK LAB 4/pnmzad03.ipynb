```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
```

implementacja funkcji g jako lambda i jej pochodnej Dg:

```python
x, h = sp.symbols("x h")

g = sp.tan(x * x * x) - 4 * sp.tan(x)

Dg = sp.diff(g, x)
```

Wykres funkcji g(x) oraz jej pochodnej w zakresie [0, 1]

```python


print("g(x) =", g)
print("Dg(x) =", Dg)

g_num = sp.lambdify(x, g, "numpy")
Dg_num = sp.lambdify(x, Dg, "numpy")

xs = np.linspace(0.5, 1.5, 500)

g_vals = g_num(xs)
Dg_vals = Dg_num(xs)

g_masked = np.ma.masked_where(np.abs(g_vals) > 100, g_vals)
Dg_masked = np.ma.masked_where(np.abs(Dg_vals) > 100, Dg_vals)

plt.figure(figsize=(10, 5))
# plt.gca().set_aspect('equal', adjustable='box')

plt.plot(xs, g_masked, label="g(x)")
plt.plot(xs, Dg_masked, label="Dg(x)")

plt.xlabel("x")
plt.ylabel("wartosc")
plt.title("Funkcja i pochodna")

plt.ylim(-100, 100)
plt.grid(True)

plt.legend()
plt.show()
```
