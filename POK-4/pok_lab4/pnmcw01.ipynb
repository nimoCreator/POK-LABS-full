# Tablice numeryczne w module obliczeniowym NumPy

### prosty import modulu oraz opis jego głównej operacji opakowania sekwencji (np. typu list) danych numerycznych w mutacyjny obiekt tablicy zapisany jako ciag bajtow w pamieci komputera i zoptymalizowany do szybkich obliczen matematycznych

```python
import numpy
help(numpy.array)
```

### tablica numeryczna zadana lista liczb calkowitych (int) lub rzeczywistych (float) lub zespolonych (complex)

```python
numpy.array([1.1,2.2,3.3], dtype=float) # odczyt wyrazenia i ewaluacja z wydrukiem wyniku jedynie w trybie interakcyjnym
```

```python
print( numpy.array([1.1,2.2,3.3], dtype=float) ) # wydruk wyniku w trybach wsadowym i interakcyjnym
```

### konwersja domyslna do wspolnego typu najbardziej zlozonego (np. complex)

```python
numpy.array([1.1,0.22j,33])
```

### tablice mozna zadac tensorem (macierza) liczb czyli lista list o regularnej wymiarowosci

```python
lst = [ [0.11,0.12,0.13], 
        [0.21,0.22,0.23], 
        [0.31,0.32,0.33] ]
```

```python
tab = numpy.array(lst)
```

### charakterystyka tablic numerycznych

```python
print("tab-type:", type(tab) ) # typ obiektu tablicy
print("itm-type:", tab.dtype ) # typ elementu tablicy
print("tab-shape:", tab.shape ) # wymiarowosc tablicy jako krotka rozmiarow dla kolejnych wymiarow tensora
print("tab-size:", tab.size ) # liczba elementow tablicy
print("itm-size:", tab.itemsize ) # rozmiar w bajtach elementu tablicy
print("tab-nbytes:", tab.nbytes ) # rozmiar w bajtach calej tablicy w pamieci
```

### dostep do danych listy list (tablica poszarpana) wymaga podwojnego operatora indeksowania z zakresem indeksow od zera do rozmiar-1

```python
print("lst[0][0] =", lst[0][0] )
```

### dostep do danych w regularnej tablicy wymaga jednego operatora z przecinkiem (tablica jest podobna do slownika o kluczu w postaci krotki indeksow)

```python
print("tab[0,0] =", tab[0,0] )
```

### inne przypadku dostepu do danych w tablicy numerycznej z uzyciem wycinkow i wielokropka

```python
print("tab[1] =", tab[1] )
print("tab[1,:] =", tab[1,:] )
print("tab[1,...] =", tab[1,...] ) # operator wielokropka uzywany jest w tablicy numerycznej podobnie jak we zwyklym slowniku
print("tab[1,2] =", tab[1,2] )
print("tab[:,2] =", tab[:,2] )
print("tab[...,2] =", tab[...,2] )
```

### dostep do calej tablicy

```python
print("tab[...] =", tab[...] )
```

### wydruk listy list zdefiniowanej w skrypcie i wystepujacej w maszynie systemu Python

```python
print("lst =", lst )
```

### wydruk listy list wydobytej z tablicy numerycznej znajdujacej sie w pamieci operacyjnej

```python
print("tab =", tab.tolist() )
```

### bezposredni wydruk tablicy numerycznej

```python
print("tab =", tab )
```

### dowiazanie wielu etykiet do jednego obiektu tablicy

```python
tab2 = tab
```

### zapis wartosci poprzez druga etykiete daje utajona mutacyjnosc obiektu tablicy

```python
tab2[1,2] = -2.3
print("tab2[1,2] =", tab2[1,2] )
print("tab[1,2] =", tab[1,2] )
```

### stad nalezy jawnie skopiowac tablice do nowego obiektu tablicy

```python
tab3 = tab2.copy()
```

### teraz modyfikacja zgodna jest z idiomem _Copy-On-Wirite_

```python
tab3[1,2] = 0.23
print("tab3[1,2] =", tab3[1,2] )
print("tab2[1,2] =", tab2[1,2] )
```

### odcztanie bajtow z pamieci tablicy co mozna np. zapisac do pliku binarnego

```python
tbin = tab3.tobytes()
print("tbin =", tbin )
```

### zapisanie bajtow do pamieci tablicy np. odczytanych z pliku binarnego

```python
tab4 = numpy.frombuffer(tbin)
print("tab4 =", tab4 )
```

### konieczne jest nastpnie odtworzenie poczatkowej wymiarowosci

```python
tab5 = tab4.reshape((3, 3)) # tworzy tablice-macierz 3 x 3
print("tab5 =", tab5 )
```
