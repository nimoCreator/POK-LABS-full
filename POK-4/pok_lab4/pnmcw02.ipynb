# Obliczenia na tablicach numerycznych modulu NumPy

```python
import numpy as np # import modulu pod skrocona nazwa
```

### utworzenie tablicy-wektora 10 liczb rzeczywistych z zakresu liczb calkowitych 0..9:1  

```python
vec1 = np.array(range(10), float)
print("vec1 =", vec1 )
```

### utworzenie tablicy-wektora 10 liczb rzeczywistych z zakresu liczb calkowitych 10..19:1  

```python
vec2 = np.arange(10,20,1, float)
print("vec2 =", vec2 )
```

### wykonanie prostych dzialan na elementach tablic

```python
print("vec1+vec2 =", vec1+vec2 )
print("vec1*vec2 =", vec1*vec2 )
print("vec1-vec2 =", vec1-vec2 )
print("vec1/vec2 =", vec1/vec2 )
```

### wykonanie prostych relacji na elementach tablic

```python
print("vec1<vec2 =", vec1<vec2 )
print("vec1>vec2 =", vec1>vec2 )
print("vec1==vec2 =", vec1==vec2 )
```

### utworzenie tablicy-wektora 10 liczb rzeczywistych pseudolosowych z przedzialu [0,1)

```python
vec = np.random.rand(10) 
print("vec =", vec )
```

### wykonanie prostych dzialan i relacji na elementach tablicy i osobnej danej skalarnej

```python
print("vec+100 =", vec+100 )
print("vec<100 =", vec<100 )
```

### wykonanie obliczen na elementach tablicy z uzyciem funkcji matematycznych

```python
print("sin(vec) =", np.sin(vec) )
print("tan(vec) =", np.tan(vec) )
print("log(vec) =", np.log(vec) )
print("sqrt(vec) =", np.sqrt(vec) )
```

### wykonanie redukcji elementow tablic do wypadkowego wyniku

```python
print("sum(vec) =", np.sum(vec) )
print("prod(vec) =", np.prod(vec) )
print("min(vec) =", np.min(vec) )
print("max(vec) =", np.max(vec) )
```

### wykonanie zlozonych obliczen na elementach tablic z redukcja wynikow

```python
print("sum( sin(pi*log(vec1+vec2)*vec) ) =", np.sum( np.sin(np.pi*np.log(vec1+vec2)*vec) ) )
```

### utworzenie tablicy-macierzy 5 x 10 liczb rzeczywistych pseudolosowych z przedzialu [0,1)

```python
mtx = np.random.rand(5,10)
print("mtx =", mtx )
print("mtx-shape:", mtx.shape)
```

### wykonanie mnozenia macierzy przez wektor

```python
mtxvec = np.dot(mtx,vec)
print("mtx . vec =", mtxvec )
print("mtxvec-shape:", mtxvec.shape)
```

### wykonanie prostych dzialan na wierszach macierzy i osobnym wektorze a nastepnie posumowanie wzdluz wymiaru drugiego czyli po wierszach macierzy co daje wynik mnozenia macierzy przez wektor

```python
mtxvec2 = np.sum(mtx*vec, 1)
print("sum(mtx*vec) =", mtxvec2 )
print("mtxvec2-shape:", mtxvec2.shape)
```

### obliczanie numerycznie pierwiastkow rownania wielomianowego 
> x^4 - 11 * x^3 + 9 * x^2 + 11 * x - 10 == 0

```python
print("roots([1, -11, 9, 11, -10]) =", np.roots([1, -11, 9, 11, -10]) )
```

### wyznaczenie pochodnej po _x_ z powyzszego wielomianu

```python
print("polyder([1, -11, 9, 11, -10]) =", np.polyder([1, -11, 9, 11, -10]) )
```

### wyznaczenie nieoznaczonej calki po _x_ z powyzszego wielomianu

```python
print("polyint([1, -11, 9, 11, -10]) =", np.polyint([1, -11, 9, 11, -10]) )
```
