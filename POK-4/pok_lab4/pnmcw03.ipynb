# Wizualizacja tablic numerycznych z uzyciem modulu MatPlotLib

### import modulu oraz opis jego głównej operacji do tworzenia wykresow danych numerycznych

```python
import numpy as np
import matplotlib.pyplot as plt
help(plt.plot)
```

### Przyklad najprostszego wykresu 2d

#### tworzy prosta tablice-wektor z liczbami calkowitymi

```python
y = np.array([2, 1, 5, 7, 4, 6, 8, 14, 10, 9, 18, 20, 22])
```

#### rysuje i wyswietla liniowo prosty wykres danych _y_ gdzie na osi OX sa indeksy liczb w tablicy a na osi OY sa liczby z tablicy

```python
plt.plot(y)
```

### Przyklad zlozonego wykresu 2d

#### tworzy tablice liczb rzeczywistych w przedziale [0,5] z liniowym podzialem na 20 punktow o rownych odstepach

```python
x = np.linspace(0, 5, 20)
print("x = ", x)
```

#### liczy funkcje exp dla kolejnych liczb z tablicy _x_

```python
y = np.exp(x)
print("y = ", y)
```

#### tworzy prosty wykres serii danych _y_(_x_) w kilku krokach:
- rysuje tytul wykresu
- rysuje czerwona linie (x,y)
- rysuje niebieskie punkty (x,y)
- rysuje nazwy osi OX i OY
- rysuje legende z etykietami
- wyswietla liniowo plansze wykresu

#### przy czym wszystkie te operacje musza byc w jednej komorce wejsciowej aby dotyczyly tego samego wykresu w komorce wyjsciowej

```python
plt.title("Seria danych")
plt.plot(x, y, color='red', linestyle='solid', label="linia")
plt.plot(x, y, color='blue', linestyle='none', marker='o', label="punkty")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
#plt.show() # pod koniec skryptu komorki wyswietlanie zachodzi niejawnie
```

### Przyklad wykresu 3d

#### tworzy plansze do rysowania wykresow

```python
fig = plt.figure() # moze tu powstac nowy kontekst wyswietlania tresci graficznych w ktorym nastapi wizualizacja figury
```

#### rysuje uklad osi XYZ z rzutowaniem 3d

```python
axs = fig.add_subplot(projection='3d')
```

#### tworzy tablice-wektory liczb rzeczywistych w zakresie od -5 do 5 z krokiem 0.15

```python
x = np.arange(-5, 5, 0.15)
print("x = ", x)
y = np.arange(-5, 5, 0.15)
print("y = ", y)
```

#### tworzy siatki 2d dla osi OX i OY

```python
X, Y = np.meshgrid(x, y)
print("X = ", X)
print("Y = ", Y)
```

#### liczy wyrazenie na elementach tablic zgodnie ze wzorem funkcji dwoch zmiennych:
> z = f(x,y) = sin( sqrt( x^2 + y^2 ) )

```python
Z = np.sin( np.sqrt(X**2 + Y**2) )
```

#### rysuje wykres powierzchniowy funkcji z wybrana mapa kolorow

```python
axs.plot_surface(X, Y, Z, cmap='Spectral')
```

#### rysuje tytul na planszy i nastepnie ja wyswietla liniowo

```python
fig.suptitle("Wykres 3d")
display(fig) # domyslnie: from IPython.display import display
```

### Przyklad zlozonego wykresu 2d z wieloma zestawami danych

#### tworzy plansze o rozmiarze 5x3 oraz rysuje na niej uklad osi XOY i wyswietla stan poczatkowy planszy

```python
fig, axs = plt.subplots(figsize=(5, 3)) # powstaje nowy kontekst wyswietlania graficznego do wizualizacji figury
```

#### zadaje ziarno dla liczb pseudolosowych

```python
np.random.seed(19680801)
```

#### tworzy tablice-wektor liczb calkowitych w zakresie od 0 do 200 z krokiem 1

```python
t = np.arange(200)
```

#### tworzy tablice 200 liczb rzeczywistych pseudolosowych o rozkladzie normalnym z srednia 0 i odchyleniem 1 a nastepnie te tablice sumuje kumulatywnie zgodnie ze wzorem: 
> bj = Sum(i=0..j) ai

```python
x = np.cumsum(np.random.randn(200))
y = np.cumsum(np.random.randn(200))
```

#### rysuje wykres danych _x_(_t_)

```python
axs.plot(t, x, label="Dane X")
```

#### rysuje wykres danych _y_(_t_)

```python
axs.plot(t, y, label="Dane Y")
```

#### rysuje nazwy osi

```python
axs.set_xlabel("Czas")
axs.set_ylabel("Polozenie")
```

#### rysuje legende z etykietami podanymi w p/w operacji plot

```python
axs.legend()
```

#### rysuje tytul nad ukladem osi XOY i wyswietla liniowo plansze

```python
axs.set_title("Bladzenie losowe")
display(fig) # ponowne utworzenie nowego kontekstu wyswietlania graficznego do wizualizacji figury
```
