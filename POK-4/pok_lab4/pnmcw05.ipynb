# Wykonywanie prostych obliczen symbolicznych z uzyciem modulu Sympy

#### importuje wszystkie elementy z modulu

```python
from sympy import *
```

#### specyfikuje abstrakcyjne symbole niewiadomych

```python
x, y, z = symbols('x y z')
```

#### specyfikuje funkcje i ciagi symboliczne

```python
f, g, h = symbols('f g h', cls=Function)
u, v, w = symbols('u v w', seq=True, cls=Function)
```

#### specyfikuje wspolczynniki i indeksy symboliczne

```python
a, b, c = symbols('a, b, c', real=True)
n, m, i = symbols('n, m, i', integer=True)
```

#### redukuje symbolicznie podane wyrazenie

```python
print("sin(x)**2 + cos(x)**2 =", simplify(sin(x)**2 + cos(x)**2) )
```

#### porownuje symbolicznie dwa wyrazenia

```python
(sin(x)**2 + cos(x)**2).equals(1)
```

#### oblicza numerycznie wyrazenia symboliczne poprzez podstawienie wartosci pod symbol zmiennej

```python
(sin(x)**2 + cos(x)**2).evalf(subs={x: 0.1})
```

#### oblicza pochodna funkcji po x

```python
diff(f(x), x)
```

```python
print("d cos(x) / dx =", diff(cos(x), x) )
```

#### pochodna trzeciego rzedu

```python
diff(f(x), x, x, x)
```

```python
print("d^3 x^4 / dx^3 =", diff(x**4, x, x, x) )
```

#### oblicza calke nieoznaczona po x

```python
integrate(f(x), x)
```

```python
print("I cos(x) dx =", integrate(cos(x), x) )
```

#### calka oznaczona po x od 0 do nieskonczonosci

```python
integrate(f(x), (x, a, b))
```

```python
print("I(0,oo) exp(-x) dx =", integrate(exp(-x), (x, 0, oo)) )
```

#### oblicza granice funkcji przy x dazacym do zera

```python
limit(f(x), x, a)
```

```python
print("lim(x->0) sin(x)/x =", limit(sin(x)/x, x, 0) )
```

#### oblicza granice przy x dazacym do nieskonczonosci

```python
limit(f(x), x, oo)
```

```python
print("lim(x->oo) (1+1/x)^x =", limit((1+1/x)**x, x, oo) )
```

#### oblicza dyskretna granice dla ciagu przy n dazacym do nieskonczonosci

```python
limit_seq(u(n), n)
```

```python
print("lim(n->oo) (1+1/n)^n =", limit_seq((1+1/n)**n, n) )
```

#### sumuje i oblicza granice sumy ciagu

```python
Sum(u(i), (i,n,m))
```

```python
print("lim(n->oo) Sum(1/i!, (i,0,n)) =", limit_seq( Sum(1/factorial(i), (i,0,n)), n) )
```

#### mnozy i oblicza granice iloczynu ciagu

```python
Product(u(i), (i,n,m))
```

```python
print("lim(n->oo) Prod((4*(i**2))/((2*i - 1)*(2*i + 1)), (i,0,n)) =", limit_seq( Product((4*(i**2))/((2*i - 1)*(2*i + 1)), (i,1,n)), n) )
```

#### rozwiazuje rownanie wzgledem x poprzez podanie lewej strony rownania z zerem po prawej

```python
print("x^2+b*x+c==0, x =", solveset(x**2+b*x+c, x) ) 
```

#### rozwiazuje rownania rekurencyjne

```python
from sympy.series.sequences import RecursiveSeq
fibo = RecursiveSeq(u(n - 1) + u(n - 2), u(n), n, [0, 1])
```
