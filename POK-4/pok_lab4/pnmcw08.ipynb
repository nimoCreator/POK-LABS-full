```python
# x = var('x')
plot(sin(2*x), x, -pi, pi)
```

```python
y = var('y')
plot3d(sin(pi*sqrt(x^2 + y^2))/sqrt(x^2+y^2), (x,-5,5), (y,-5,5))
```

```python
plot(lambda x:fibonacci(round(x)), (x,1,10))
```

```python
show(sum(circle((i,0), i, hue=sin(i/10)) for i in [10,9.9,..,0]))
```

```python
g1 = plot(x^2*sin(1/x), (x, -2, 2), axes_labels=['$x$', '$y$'])
g2 = plot(x^2*sin(1/x), (x, -0.3, 0.3), axes_labels=['$x$', '$y$'],frame=True)
g1.inset(g2, pos=(0.15, 0.7, 0.25, 0.25))
```

```python
y(x) = x*sin(x^2)
v = [(x, y(x)) for x in [-3,-2.95,..,3]]
show(points(v, rgbcolor=(0.2,0.6, 0.1), pointsize=30) + plot(spline(v), -3.1, 3))
```
