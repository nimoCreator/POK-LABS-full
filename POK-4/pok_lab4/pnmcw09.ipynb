# Wykonywanie prostych operacji matematycznych w SageMath

```python
integral(x*sin(x^2), x)
```

```python
integral(x/(x^2+1), x, 0, 1)
```

```python
f = 1/((1+x)*(x-1))
f.partial_fraction(x)
```

```python
t = var('t')    # define a variable t

x = function('x')(t)   # define x to be a function of that variable

DE = diff(x, t) + x - 1

desolve(DE, [x,t])
```

```python
s = var("s")

t = var("t")

f = t^2*exp(t) - sin(t)

f.laplace(t,s)
```

```python
t,x,y = PolynomialRing(RealField(10),3,"txy").gens()

f = y; g = -x - y * t

eulers_method_2x2(f,g, 0, 1, 0, 1/4, 1)
```

```python
x = polygen(QQ, 'x')

chebyshev_U(2,x)
```

```python
bessel_I(1,1).n(250)
```

```python
theta = var('theta')

solve(cos(theta)==sin(theta), theta)
```

```python
phi = var('phi')

find_root(cos(phi)==sin(phi),0,pi/2)
```

```python
var('x y p q')
eq1 = p+q==9

eq2 = q*y+p*x==-6

eq3 = q*y^2+p*x^2==24

solve([eq1,eq2,eq3,p==1],p,q,x,y)
```
